```python
import matplotlib.pyplot as plt
```

```python
def midpoint_circle(xc, yc, r):
    x = 0
    y = r
    p = 1 - r

    points = []

    def plot_symmetric_points(x, y):
        symmetric = [
            (xc + x, yc + y),
            (xc - x, yc + y),
            (xc + x, yc - y),
            (xc - x, yc - y),
            (xc + y, yc + x),
            (xc - y, yc + x),
            (xc + y, yc - x),
            (xc - y, yc - x)
        ]
        points.extend(symmetric)

    plot_symmetric_points(x, y)

    while x < y:
        x += 1

        if p < 0:
            p += 2 * x + 1
        else:
            y -= 1
            p += 2 * x - 2 * y + 1

        plot_symmetric_points(x, y)

    return points
```

```python
face_x = int(input("Enter face center X: "))
face_y = int(input("Enter face center Y: "))
face_r = int(input("Enter face radius: "))
```

```python
face = midpoint_circle(face_x, face_y, face_r)
```

```python
eye_r = face_r // 8
```

```python
left_eye = midpoint_circle(
    face_x - face_r // 3,
    face_y + face_r // 3,
    eye_r
)
```

```python
right_eye = midpoint_circle(
    face_x + face_r // 3,
    face_y + face_r // 3,
    eye_r
)
```

```python
smile = midpoint_circle(
    face_x,
    face_y - face_r // 6,
    face_r // 2
)
```

```python
smile_arc = []
```

```python
for x, y in smile:
    if y < (face_y - face_r // 6):
        smile_arc.append((x, y))
```

```python
all_points = face + left_eye + right_eye + smile_arc
```

```python
x_vals = [p[0] for p in all_points]
y_vals = [p[1] for p in all_points]
```

```python
plt.figure(figsize=(7, 7))
```

```python
plt.scatter(x_vals, y_vals, marker='s')
```

```python
plt.title("Smiley Face Using Midpoint Circle Algorithm")
```

```python
plt.axis('equal')
```
