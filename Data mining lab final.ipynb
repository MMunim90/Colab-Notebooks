```python
# ============================================================
# STEP 1: Import Required Libraries
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score


# ============================================================
# STEP 2: Upload CSV File from Your Computer
# ============================================================

from google.colab import files

uploaded = files.upload()


# ============================================================
# STEP 3: Check Uploaded File Name
# ============================================================

print("Uploaded file(s):")

for filename in uploaded.keys():
    print(filename)


# ============================================================
# STEP 4: Get File Name Automatically
# ============================================================

filename = list(uploaded.keys())[0]

print("\nSelected File:", filename)


# ============================================================
# STEP 5: Load CSV Dataset
# ============================================================

df = pd.read_csv(filename)


# ============================================================
# STEP 6: Display First 5 Rows
# ============================================================

print("\nFirst 5 Rows:")
display(df.head())


# ============================================================
# STEP 7: Display Last 5 Rows
# ============================================================

print("\nLast 5 Rows:")
display(df.tail())


# ============================================================
# STEP 8: Check Dataset Shape
# ============================================================

print("\nDataset Shape:")
print(df.shape)

print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])


# ============================================================
# STEP 9: Check Column Names
# ============================================================

print("\nColumn Names:")
print(df.columns.tolist())


# ============================================================
# STEP 10: Check Data Types
# ============================================================

print("\nData Types:")
print(df.dtypes)


# ============================================================
# STEP 11: Dataset Information
# ============================================================

print("\nDataset Information:")
df.info()


# ============================================================
# STEP 12: Check Missing Values
# ============================================================

print("\nMissing Values:")
print(df.isnull().sum())


# ============================================================
# STEP 13: Check Duplicate Rows
# ============================================================

print("\nNumber of Duplicate Rows:")
print(df.duplicated().sum())


# ============================================================
# STEP 14: Statistical Summary
# ============================================================

print("\nStatistical Summary:")
display(df.describe())


# ============================================================
# STEP 15: Display Complete Dataset
# ============================================================

print("\nComplete Dataset:")
display(df)


# ============================================================
# STEP 16: EDA - Scatter Plot
# Visualize Underlying Structure / Shape of Data
# ============================================================

plt.figure(figsize=(10, 7))

plt.scatter(
    df["Annual Income (k$)"],
    df["Spending Score (1-100)"],
    s=80,
    alpha=0.7
)

plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score (1-100)")
plt.title("Customer Segmentation: Annual Income vs Spending Score")

plt.grid(True, alpha=0.3)

plt.show()


# ============================================================
# STEP 17: Select Appropriate Numerical Features for Clustering
# ============================================================

X = df[[
    "Annual Income (k$)",
    "Spending Score (1-100)"
]]

print("\nSelected Numerical Features for Clustering:")
display(X.head())

print("\nSelected Feature Names:")
print(X.columns.tolist())

print("\nShape of Clustering Data:")
print(X.shape)


# ============================================================
# STEP 18: Standardize the Data Using StandardScaler
# ============================================================

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)


# ============================================================
# STEP 19: Convert Scaled Data into DataFrame
# ============================================================

X_scaled_df = pd.DataFrame(
    X_scaled,
    columns=X.columns
)


# ============================================================
# STEP 20: Display Standardized Data
# ============================================================

print("\nStandardized Data:")
display(X_scaled_df.head())


# ============================================================
# STEP 21: Check Mean and Standard Deviation
# ============================================================

print("\nMean of Standardized Features:")
print(X_scaled_df.mean())

print("\nStandard Deviation of Standardized Features:")
print(X_scaled_df.std())


# ============================================================
# STEP 22: Visualize Standardized Data
# ============================================================

plt.figure(figsize=(10, 7))

plt.scatter(
    X_scaled_df["Annual Income (k$)"],
    X_scaled_df["Spending Score (1-100)"],
    s=80,
    alpha=0.7
)

plt.xlabel("Standardized Annual Income")
plt.ylabel("Standardized Spending Score")
plt.title("Standardized Data for Customer Clustering")

plt.grid(True, alpha=0.3)

plt.show()


# ============================================================
# STEP 23: Find Appropriate Number of Clusters
# Using Elbow Method
# ============================================================

inertia = []

K_range = range(2, 11)

for k in K_range:

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X_scaled)

    inertia.append(kmeans.inertia_)


# ============================================================
# STEP 24: Plot Elbow Method
# ============================================================

plt.figure(figsize=(10, 7))

plt.plot(
    K_range,
    inertia,
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method for Selecting Optimal Number of Clusters")

plt.grid(True, alpha=0.3)

plt.show()


# ============================================================
# STEP 25: Calculate Silhouette Scores
# ============================================================

silhouette_scores = []

for k in K_range:

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(X_scaled)

    score = silhouette_score(
        X_scaled,
        labels
    )

    silhouette_scores.append(score)

    print(
        f"Number of Clusters = {k}, "
        f"Silhouette Score = {score:.4f}"
    )


# ============================================================
# STEP 26: Plot Silhouette Scores
# ============================================================

plt.figure(figsize=(10, 7))

plt.plot(
    K_range,
    silhouette_scores,
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score for Different Numbers of Clusters")

plt.grid(True, alpha=0.3)

plt.show()


# ============================================================
# STEP 27: Select Number of Clusters
# ============================================================

# Based on the Elbow Method and Silhouette Score,
# we select 5 clusters for customer segmentation.

optimal_k = 5

print("\nSelected Number of Clusters:", optimal_k)


# ============================================================
# STEP 28: Choose Appropriate Affinity / Distance Measure
# ============================================================

# K-Means uses Euclidean distance to measure
# the distance between data points and cluster centroids.

affinity = "Euclidean Distance"

print("Selected Affinity / Distance Measure:", affinity)


# ============================================================
# STEP 29: Apply K-Means Clustering
# ============================================================

kmeans = KMeans(
    n_clusters=optimal_k,
    random_state=42,
    n_init=10
)

cluster_labels = kmeans.fit_predict(X_scaled)


# ============================================================
# STEP 30: Add Cluster Labels to Original Dataset
# ============================================================

df["Cluster"] = cluster_labels


# ============================================================
# STEP 31: Display Clustered Dataset
# ============================================================

print("\nDataset with Cluster Labels:")
display(df.head(20))


# ============================================================
# STEP 32: Calculate Final Silhouette Score
# ============================================================

final_silhouette_score = silhouette_score(
    X_scaled,
    cluster_labels
)

print(
    "\nFinal Silhouette Score:",
    round(final_silhouette_score, 4)
)


# ============================================================
# STEP 33: Get Cluster Centers
# ============================================================

cluster_centers_scaled = kmeans.cluster_centers_


# ============================================================
# STEP 34: Convert Cluster Centers Back to Original Scale
# ============================================================

cluster_centers_original = scaler.inverse_transform(
    cluster_centers_scaled
)


# ============================================================
# STEP 35: Display Cluster Centers
# ============================================================

cluster_centers_df = pd.DataFrame(
    cluster_centers_original,
    columns=[
        "Annual Income (k$)",
        "Spending Score (1-100)"
    ]
)

print("\nCluster Centers:")
display(cluster_centers_df)


# ============================================================
# STEP 36: Visualize Final Customer Clusters
# ============================================================

plt.figure(figsize=(10, 7))

plt.scatter(
    X_scaled[:, 0],
    X_scaled[:, 1],
    c=cluster_labels,
    s=80,
    alpha=0.7
)

plt.scatter(
    cluster_centers_scaled[:, 0],
    cluster_centers_scaled[:, 1],
    marker="X",
    s=250,
    edgecolors="black",
    linewidths=2,
    label="Cluster Centers"
)

plt.xlabel("Standardized Annual Income")
plt.ylabel("Standardized Spending Score")
plt.title("Customer Segmentation Using K-Means Clustering")

plt.legend()
plt.grid(True, alpha=0.3)

plt.show()


# ============================================================
# STEP 37: Display Number of Customers in Each Cluster
# ============================================================

print("\nNumber of Customers in Each Cluster:")

cluster_counts = df["Cluster"].value_counts().sort_index()

print(cluster_counts)
```
