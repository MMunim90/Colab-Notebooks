```python
import matplotlib.pyplot as plt
```

```python
x1 = int(input("Enter x1: "))
y1 = int(input("Enter y1: "))
x2 = int(input("Enter x2: "))
y2 = int(input("Enter y2: "))
```

```python
dx = x2 - x1
dy = y2 - y1
```

```python
p = 2 * dy - dx
```

```python
x = x1
y = y1
```

```python
x_points = []
y_points = []
```

```python
while x <= x2:
    x_points.append(x)
    y_points.append(y)

    x = x + 1

    if p < 0:
        p = p + 2 * dy
    else:
        y = y + 1
        p = p + 2 * dy - 2 * dx
```

```python
plt.figure(figsize=(6,6))
plt.plot(x_points, y_points, marker='o')
```

```python
plt.grid(True)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Bresenham Line Drawing Algorithm")
```

```python
plt.show()
```
