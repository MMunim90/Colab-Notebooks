```python
import matplotlib.pyplot as plt

def midpoint_circle(xc, yc, r):
    x = 0
    y = r
    p = 1 - r

    points = []

    def plot_points(x, y):
        pts = [
            (xc+x, yc+y), (xc-x, yc+y),
            (xc+x, yc-y), (xc-x, yc-y),
            (xc+y, yc+x), (xc-y, yc+x),
            (xc+y, yc-x), (xc-y, yc-x)
        ]
        points.extend(pts)

    plot_points(x, y)

    while x < y:
        x += 1

        if p < 0:
            p += 2*x + 1
        else:
            y -= 1
            p += 2*x - 2*y + 1

        plot_points(x, y)

    return points

xc, yc, r = 0, 0, 50

pts = midpoint_circle(xc, yc, r)

x = [p[0] for p in pts]
y = [p[1] for p in pts]

plt.figure(figsize=(6,6))
plt.scatter(x, y, s=10)
plt.axis('equal')
plt.grid(True)
plt.title("Midpoint Circle Drawing")
plt.show()
```
